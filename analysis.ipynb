```python
import sqlite3
import pandas as pd

# 1. Reload our clean dataset back into Python's memory (Fixed spelling here!)
df = pd.read_csv("cleaned_agriculture_data.csv")

# 2. Open a new database file named agri_database
conn = sqlite3.connect("agri_database.db")

# 3. Put all your rows into an SQL table called crop_data
df.to_sql("crop_data", conn, if_exists="replace", index=False)

print("SQL Database successfully created! Table 'crop_data' is ready.")
```

```python
# Let's select ALL columns using * to see the exact headers SQL created
query = "SELECT * FROM crop_data LIMIT 5;"

# Execute the query and display the table
sql_result = pd.read_sql_query(query, conn)
sql_result
```

```python
# SQL query to find total production grouped by each State
query = """
SELECT State, SUM(Production) AS Total_Production
FROM crop_data
GROUP BY State
ORDER BY Total_Production DESC;
"""

# Run the query
state_production = pd.read_sql_query(query, conn)
state_production
```

```python
# SQL query to count how many times each crop appears in our data
crop_query = """
SELECT Crop, COUNT(*) AS Total_Occurrences
FROM crop_data
GROUP BY Crop
ORDER BY Total_Occurrences DESC
LIMIT 10;
"""

# Run the query and show the top 10 most frequent crops
crop_counts = pd.read_sql_query(crop_query, conn)
crop_counts
```
